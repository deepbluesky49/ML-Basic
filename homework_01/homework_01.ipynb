## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
# ВАШ КОД ЗДЕСЬ
num_arab = int(input())

if num_arab < 1 or num_arab > 3999:
    print("Ошибка")
else:
    thousands = ["", "M", "MM", "MMM"]
    hundreds = ["", "C", "CC", "CCC", "CD", "D", "DC", "DCC", "DCCC", "CM"]
    tens = ["", "X", "XX", "XXX", "XL", "L", "LX", "LXX", "LXXX", "XC"]
    ones = ["", "I", "II", "III", "IV", "V", "VI", "VII", "VIII", "IX"]

    num_roman = (thousands[num_arab // 1000] +
             hundreds[(num_arab % 1000) // 100] +
             tens[(num_arab % 100) // 10] +
             ones[num_arab % 10])
    print(num_roman)
```
